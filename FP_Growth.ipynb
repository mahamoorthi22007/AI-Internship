```python
import pandas as pd
df=pd.read_csv('/content/Market_Basket_Optimisation.csv')
```

```python
transactions=[]
for i in range(len(df)):
  transactions.append([str(df.values[i,j]) for j in range(0,20) if str(df.values[i,j])!='nan'])
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth
```

```python
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df = pd.DataFrame(te_array, columns=te.columns_)

# Apply FP-Growth
frequent_itemsets = fpgrowth(
    df,
    min_support=0.4,
    use_colnames=True
)

print(frequent_itemsets)
```

```python
!pip install mlxtend

from mlxtend.frequent_patterns import fpgrowth, association_rules
from mlxtend.preprocessing   import TransactionEncoder
import pandas as pd

# Step 1 — Encode transactions into one-hot format
te      = TransactionEncoder()
te_arr  = te.fit(transactions).transform(transactions)
df_fp   = pd.DataFrame(te_arr, columns=te.columns_)

df_fp.head()
# Step 2 — Get frequent itemsets
frequent_itemsets = fpgrowth(
    df_fp,
    min_support     = 0.003,
    use_colnames    = True,
    max_len         = 3       # max_length
)

frequent_itemsets
# Step 3 — Generate association rules
rules = association_rules(
    frequent_itemsets,
    metric          = 'lift',
    min_threshold   = 2       # min_lift
)

# Step 4 — Filter and clean
rules = rules[
    (rules['confidence'] >= 0.4) &   # min_confidence
    (rules['lift']       >= 2)
]

# Step 5 — Convert frozensets to readable strings
rules['antecedents'] = rules['antecedents'].apply(lambda x: ', '.join(sorted(x)))
rules['consequents'] = rules['consequents'].apply(lambda x: ', '.join(sorted(x)))

# Step 6 — Select and rename columns
output_DataFrame = rules[[
    'antecedents',
    'consequents',
    'support',
    'confidence',
    'lift'
]].rename(columns={
    'antecedents' : 'Left Hand Side',
    'consequents' : 'Right Hand Side',
    'support'     : 'Support',
    'confidence'  : 'Confidence',
    'lift'        : 'Lift'
}).reset_index(drop=True)

output_DataFrame.sort_values(by='Lift', ascending=False).reset_index(drop=True)
```
