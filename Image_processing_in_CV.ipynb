```python
rotation_90_c=cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(rotation_90_c)
```

```python
rotation_90_ac=cv2.rotate(img,cv2.ROTATE_90_COUNTERCLOCKWISE)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(rotation_90_ac)
```

```python
rotation_180=cv2.rotate(img,cv2.ROTATE_180)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(rotation_180)
```

```python
flip=cv2.flip(img,1)
#1-vertical
#0-horizontal
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(flip)
```

```python
resize=cv2.resize(img,(300,300))
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(resize)
```

```python
blur=cv2.blur(img,(10,50))
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(blur)
```
