```python
from transformers import pipeline
```

```python
pipe = pipeline("text-generation", model="sapientinc/HRM-Text-1B")
pipe("I don't know how to solve this",max_length=40,num_return_sequences=4)
```

```python
pipe=pipeline('fill-mask',model="Davlan/xlm-roberta-base-finetuned-swahili")
```

```python
pipe("I eat <mask> at night",top_k=3)
```

```python
pipe = pipeline("question-answering", model="google-bert/bert-large-cased-whole-word-masking-finetuned-squad")
```

```python
pipe("which states are belongs to INDIA",context='India has multiple states such as New Delhi,Tamilnadu,Kerala')
```

```python
pipe = pipeline("summarization", model="LeoCordoba/beto2beto-cc-news-es-titles")
```

```python
pipe("""
A paragraph is a distinct, self-contained section of writing that groups sentences to support one central idea or controlling topic. It typically consists of a topic sentence introducing the main point, several supporting sentences that provide evidence or details, and a concluding sentence that summarizes the thought.""")
```
