```python
import pandas as pd
```

```python
df=pd.read_csv('/content/fake_news_dataset.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
l=LabelEncoder()
```

```python
df['label']=l.fit_transform(df['label'])
```

```python
x=df['text']
y=df['label']
```

```python
y.value_counts()
```

```python
import tensorflow as tf
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
from tensorflow.keras.layers import LSTM,Dense,Bidirectional,Embedding
```

```python
voc_size=5000
```

```python
msg=x.copy()
```

```python
import nltk
import re
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()#for stemming purpose -normalize the words(run,ran,running)
```

```python
corpus=[]
for i in range(len(msg)):
  review=re.sub('[^a-zA-Z]',' ',msg[i])
  review=review.lower()
  review=review.split()
  review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
  review=' '.join(review)
  corpus.append(review)
```

```python
corpus
```

```python
onehot=[one_hot(words,voc_size) for words in corpus]
print(onehot)
```

```python
#Embedding
sen_len=20
embedding=pad_sequences(onehot,padding='pre',maxlen=sen_len)
print(embedding)
```

```python
embedding_vector_features=40
model=Sequential()
model.add(Embedding(voc_size,embedding_vector_features,input_length=sen_len))
model.add(Bidirectional(LSTM(100)))
model.add(Dense(1,activation='sigmoid'))
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
print(model.summary())
```

```python
import numpy as np
in_x=np.array(embedding)
in_y=np.array(y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(in_x,in_y,test_size=0.33,random_state=42)
```

```python
model.fit(x_train,y_train,validation_data=(x_test,y_test),epochs=10,batch_size=32)
```
